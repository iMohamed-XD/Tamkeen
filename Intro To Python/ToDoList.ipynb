**Variables**

```python
running = True
Tasks = []
```

**MAIN LOOP**

```python
print('====TODO APP====')
while running:
    choice = int(input("""
    1 - Show Tasks
    2 - add Task
    3 - Delete Task
    4 - N# of Tasks
    5 - Search Tasks
    6 - Exit
    CHOOSE a number 1 to 6"""))
    match choice:
        case 1:
            for i, t in enumerate(Tasks):
                print(f"{i+1} - {t}")
        case 2:
            task = input("Task name: ")
            if task and task != "" and task != " " and task not in Tasks:
                Tasks.append(task)
            else:
                print("Invalid task name or task already exists")
        case 3:
            task = input("Task name: ")
            if task in Tasks:
                Tasks.remove(task)
            else:
                print("Task not Found")
        case 4:
            print(f"Number of Tasks is = {len(Tasks)}")
        case 5:
            task = input("Task name: ")
            if task in Tasks:
                index = Tasks.index(task)
                print(f"Found {task} at index N# {index}")
            else:
                print("Task not Found")
        case 6:
            print("====STOPPED====")
            running = False
        case _:
            print("Invalid choice!!")
```
