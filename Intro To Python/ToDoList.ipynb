**Variables**

```python
running: bool = True
Tasks: list[list] = []
priorities: tuple[str] = ("High", "Medium", "Low")
status: tuple[str] = ("Pending", "Done")
```

```python
def choice():
    return int(input("""
    1 - Show Active Tasks
    2 - Add Task
    3 - Delete Task
    4 - Mark Task as Done
    5 - Search Tasks
    6 - Show Number of Active Tasks
    7 - Show Task Statistics
    8 - Delete Completed Tasks
    9 - Exit
    CHOOSE a number 1 to 9"""))
```

**MAIN Functions**

```python
def show_active_tasks():
    priority = input("Enter priority (High, Medium, Low) or nothing for all: ")
    if priority == "":
        active = [t for t in Tasks if t[-1] == status[0]]
    elif priority in priorities:
        active = [t for t in Tasks if t[-1] == status[0] and t[1] == priority]
    else:
        print("Invalid choice, showing all active tasks.")
        active = [t for t in Tasks if t[-1] == status[0]]
    if not active:
        print("No active tasks.")
        return
    print("===== ACTIVE TASKS =====")
    for i, t in enumerate(active, 1):
        print(f"{i} - {t[0]}\n   Priority: {t[1]}")
```

```python
def add_task(task_name, priority = "Low"):
    task = [task_name, priority, status[0]]
    if task and task[0].strip() and task not in Tasks:
        Tasks.append(task)
    else:
        print("Invalid task name or task already exists")
```

```python
def delete_task(task_name):
    for task in Tasks:
        if task[0] == task_name:
            Tasks.remove(task)
            return
    print("Task not found")
```

```python
def delete_completed_tasks():
    global Tasks
    before = len(Tasks)
    Tasks = [t for t in Tasks if t[-1] != status[-1]]
    print(f"{before - len(Tasks)} task(s) deleted successfully!")
```

```python
def mark_done(task_name):
    for task in Tasks:
        if task[0] == task_name:
            task[-1] = status[-1]
            return
    print("Task not found")
```

```python
def number_of_active_tasks():
    count = 0
    for task in Tasks:
        if task[-1] == status[0]:
            count += 1
    return count
```

```python
def task_statistics():
    total_tasks = len(Tasks)
    active_tasks = number_of_active_tasks()
    completed_tasks = len([k for k in Tasks if k[-1] == status[-1]])
    high_priority_tasks = len([k for k in Tasks if k[1] == priorities[0]])
    medium_priority_tasks = len([k for k in Tasks if k[1] == priorities[1]])
    low_priority_tasks = len([k for k in Tasks if k[1] == priorities[-1]])
    print(f"""
    Total tasks = {total_tasks}
    Active tasks = {active_tasks}
    Completed tasks = {completed_tasks}
    
    High priority tasks = {high_priority_tasks}
    Medium priority tasks = {medium_priority_tasks}
    Low priority tasks = {low_priority_tasks}
""")
```

```python
def search(keyword, priority=None):
    keyword = keyword.lower()
    matches = [t for t in Tasks if keyword in t[0].lower() and t[-1] == status[0] and (priority is None or t[1] == priority)]    
    if not matches:
        print("Task not Found")
        return
    print("===== SEARCH RESULTS =====")
    for i, t in enumerate(matches, 1):
        print(f"{i}. {t[0]}")
```

```python
def stop():
    global running
    running = False
```

**MAIN LOOP**

```python
print('====TODO APP====')
while running:
    choicee = choice()
    match choicee:
        case 1:
            show_active_tasks()
        case 2:
            task_name = input("Task name: ")
            priority = input("Task priority (High, Medium, Low): ")
            if priority not in priorities:
                print("Invalid priority. Setting to Low by default.")
                priority = "Low"
            add_task(task_name=task_name, priority=priority)
        case 3:
            task = input("Task name: ")
            delete_task(task)
        case 4:
            task_name = input("Task name: ")
            mark_done(task_name=task_name)
        case 5:
            keyword = input("Enter a Keyword: ")
            priority = input("Enter a priority (high, medium, low): ")
            search(keyword=keyword, priority=priority)
        case 6:
            print(f"Number of active tasks is = {number_of_active_tasks()}")
        case 7:
            task_statistics()
        case 8:
            delete_completed_tasks()
        case 9:
            print("====STOPPED====")
            stop()
        case _:
            print("Invalid choice!!")
```
