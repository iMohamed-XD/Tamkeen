**Variables**

```python
running: bool = True
Tasks: list[str] = []
```

```python
def choice():
    return int(input("""
    1 - Show Tasks
    2 - add Task
    3 - Delete Task
    4 - N# of Tasks
    5 - Search Tasks
    6 - Exit
    CHOOSE a number 1 to 6"""))
```

**MAIN Functions**

```python
def show_tasks():
    for i, t in enumerate(Tasks):
        print(f"{i+1} - {t}")
```

```python
def add_task(task):
    if task and task != "" and task != " " and task not in Tasks:
        Tasks.append(task)
    else:
        print("Invalid task name or task already exists")
```

```python
def delete_task(task):
    if task in Tasks:
        Tasks.remove(task)
    else:
        print("Task not found")
```

```python
def number_of_tasks():
    return len(Tasks)
```

```python
def search(task):
    if task in Tasks:
        index = Tasks.index(task)
        print(f"Found {task} at index N# {index}")
    else:
        print("Task not Found")
```

```python
def stop():
    global running
    running = False
```

**MAIN LOOP**

```python
print('====TODO APP====')
while running:
    choice = choice()
    match choice:
        case 1:
            show_tasks()
        case 2:
            task = input("Task name: ")
            add_task(task)
        case 3:
            task = input("Task name: ")
            delete_task(task)
        case 4:
            print(f"Number of Tasks is = {number_of_tasks()}")
        case 5:
            task = input("Task name: ")
            search(task)
        case 6:
            print("====STOPPED====")
            stop()
        case _:
            print("Invalid choice!!")
```
